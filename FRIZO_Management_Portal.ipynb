```python
# FRIZO Management Portal

products = [
    {"id": 1, "name": "Classic Plain Fries", "category": "Plain Fries", "price": 250},
    {"id": 2, "name": "Spicy Curly Fries", "category": "Curly Fries", "price": 350},
    {"id": 3, "name": "Crispy Waffle Fries", "category": "Waffle Fries", "price": 400},
    {"id": 4, "name": "Cheesy Loaded Fries", "category": "Loaded Fries", "price": 550},
    {"id": 5, "name": "Supreme Pizza Fries", "category": "Pizza Fries", "price": 600},
    {"id": 6, "name": "Masala Tornado Fries", "category": "Tornado Fries", "price": 450},
]
campaigns = []
customers = []
promotions = []

categories = ["Plain Fries", "Curly Fries", "Waffle Fries",
              "Loaded Fries", "Pizza Fries", "Tornado Fries"]
platforms = ["Instagram", "Facebook", "TikTok", "YouTube", "In-store"]
statuses = ["Planned", "Active", "Completed", "Paused"]

def get_number(message):
    while True:
        try:
            number = float(input(message))
            if number < 0:
                print("Enter a number that is 0 or more.")
            else:
                return number
        except ValueError:
            print("Invalid input. Enter a number.")

def get_integer(message):
    while True:
        try:
            number = int(input(message))
            if number < 0:
                print("Enter a number that is 0 or more.")
            else:
                return number
        except ValueError:
            print("Invalid input. Enter a whole number.")

def get_text(message):
    text = input(message).strip()
    while text == "":
        print("This cannot be empty.")
        text = input(message).strip()
    return text

def choose(title, options):
    print("\n" + title)
    for i in range(len(options)):
        print(i + 1, ".", options[i])
    choice = get_integer("Choose: ")
    while choice < 1 or choice > len(options):
        print("Choose a number from the list.")
        choice = get_integer("Choose: ")
    return options[choice - 1]

def next_id(records):
    biggest = 0
    for record in records:
        if record["id"] > biggest:
            biggest = record["id"]
    return biggest + 1

def find_by_id(records, record_id):
    for record in records:
        if record["id"] == record_id:
            return record
    return None

def login():
    print("\n===== FRIZO ADMIN LOGIN =====")
    attempts = 3
    while attempts > 0:
        username = input("Username: ")
        password = input("Password: ")
        if username == "admin" and password == "frizo123":
            print("Login successful!")
            return True
        attempts -= 1
        print("Wrong login. Attempts left:", attempts)
    print("Access denied.")
    return False

def show_dashboard():
    active = 0
    for c in campaigns:
        if c["status"] == "Active":
            active += 1
    print("\n===== FRIZO DASHBOARD =====")
    print("Products:", len(products))
    print("Campaigns:", len(campaigns), "(Active:", active, ")")
    print("Customers:", len(customers))
    print("Promo codes:", len(promotions))

def show_products():
    print("\n===== PRODUCTS =====")
    for p in products:
        print(p["id"], "|", p["name"], "|", p["category"], "| Rs.", p["price"])

def add_product():
    name = get_text("Product name: ")
    category = choose("Select category:", categories)
    price = get_number("Price (Rs.): ")
    products.append({"id": next_id(products), "name": name,
                     "category": category, "price": price})
    print("Product added!")

def search_product():
    word = get_text("Search name or category: ").lower()
    found = False
    for p in products:
        if word in p["name"].lower() or word in p["category"].lower():
            print(p["id"], "|", p["name"], "|", p["category"], "| Rs.", p["price"])
            found = True
    if not found:
        print("No product found.")

def update_product():
    show_products()
    p = find_by_id(products, get_integer("Product ID to update: "))
    if p is None:
        print("Product not found.")
        return
    p["name"] = get_text("New name: ")
    p["category"] = choose("New category:", categories)
    p["price"] = get_number("New price: ")
    print("Product updated!")

def show_campaigns():
    print("\n===== CAMPAIGNS =====")
    if len(campaigns) == 0:
        print("No campaigns yet.")
    for c in campaigns:
        print("-" * 40)
        print("ID:", c["id"])
        print("Name:", c["name"])
        print("Product:", c["product"])
        print("Platform:", c["platform"])
        print("Budget: Rs.", c["budget"])
        print("Status:", c["status"])

def add_campaign():
    show_products()
    product = find_by_id(products, get_integer("Product ID for campaign: "))
    if product is None:
        print("Product not found.")
        return
    name = get_text("Campaign name: ")
    platform = choose("Select platform:", platforms)
    budget = get_number("Budget (Rs.): ")
    campaigns.append({"id": next_id(campaigns), "name": name,
                      "product": product["name"], "platform": platform,
                      "budget": budget, "status": "Planned",
                      "reach": 0, "clicks": 0, "conversions": 0,
                      "spend": 0, "revenue": 0})
    print("Campaign created!")

def search_campaign():
    word = get_text("Search campaign name or product: ").lower()
    found = False
    for c in campaigns:
        if word in c["name"].lower() or word in c["product"].lower():
            print(c["id"], "|", c["name"], "|", c["product"], "|", c["status"])
            found = True
    if not found:
        print("No campaign found.")

def update_campaign():
    show_campaigns()
    c = find_by_id(campaigns, get_integer("Campaign ID to update: "))
    if c is None:
        print("Campaign not found.")
        return
    c["status"] = choose("Select new status:", statuses)
    print("Campaign updated!")

def delete_campaign():
    show_campaigns()
    c = find_by_id(campaigns, get_integer("Campaign ID to delete: "))
    if c is None:
        print("Campaign not found.")
        return
    campaigns.remove(c)
    print("Campaign deleted!")

def record_performance():
    show_campaigns()
    c = find_by_id(campaigns, get_integer("Campaign ID: "))
    if c is None:
        print("Campaign not found.")
        return
    c["reach"] = get_integer("Reach: ")
    c["clicks"] = get_integer("Clicks: ")
    c["conversions"] = get_integer("Conversions: ")
    c["spend"] = get_number("Spend (Rs.): ")
    c["revenue"] = get_number("Revenue (Rs.): ")
    print("Performance recorded!")

def add_customer():
    name = get_text("Customer name: ")
    phone = get_text("Phone: ")
    favourite = choose("Favourite fries:", categories)
    customers.append({"id": next_id(customers), "name": name,
                      "phone": phone, "favourite": favourite})
    print("Customer registered!")

def show_customers():
    print("\n===== CUSTOMERS =====")
    if len(customers) == 0:
        print("No customers yet.")
    for c in customers:
        print(c["id"], "|", c["name"], "|", c["phone"], "|", c["favourite"])

def search_customer():
    word = get_text("Search name or phone: ").lower()
    found = False
    for c in customers:
        if word in c["name"].lower() or word in c["phone"]:
            print(c["id"], "|", c["name"], "|", c["phone"], "|", c["favourite"])
            found = True
    if not found:
        print("No customer found.")

def add_promotion():
    code = get_text("Promo code (e.g. FRIZO20): ").upper()
    for p in promotions:
        if p["code"] == code:
            print("Code already exists.")
            return
    discount = get_number("Discount %: ")
    if discount <= 0 or discount > 100:
        print("Discount must be between 1 and 100.")
        return
    minimum = get_number("Minimum purchase (Rs.): ")
    promotions.append({"id": next_id(promotions), "code": code,
                       "discount": discount, "minimum": minimum,
                       "active": True})
    print("Promo code created!")

def show_promotions():
    print("\n===== PROMOTIONS =====")
    if len(promotions) == 0:
        print("No promo codes yet.")
    for p in promotions:
        if p["active"]:
            state = "Active"
        else:
            state = "Inactive"
        print(p["id"], "|", p["code"], "|", p["discount"], "% | Min Rs.",
              p["minimum"], "|", state)

def toggle_promotion():
    show_promotions()
    p = find_by_id(promotions, get_integer("Promotion ID: "))
    if p is None:
        print("Promotion not found.")
        return
    p["active"] = not p["active"]
    print("Status changed!")

def show_report():
    print("\n===== MARKETING REPORT =====")
    if len(campaigns) == 0:
        print("No campaign data yet.")
        return

    budget = spend = revenue = reach = clicks = conversions = 0
    for c in campaigns:
        budget += c["budget"]
        spend += c["spend"]
        revenue += c["revenue"]
        reach += c["reach"]
        clicks += c["clicks"]
        conversions += c["conversions"]

    print("Total budget: Rs.", budget)
    print("Total spend: Rs.", spend)
    print("Total revenue: Rs.", revenue)
    print("Total reach:", reach)
    print("Total clicks:", clicks)
    print("Total conversions:", conversions)

    if reach > 0:
        print("Click rate:", round(clicks / reach * 100, 2), "%")
    if clicks > 0:
        print("Conversion rate:", round(conversions / clicks * 100, 2), "%")
    if spend > 0:
        print("ROMI:", round((revenue - spend) / spend * 100, 2), "%")

    print("\n--- Per campaign ---")
    for c in campaigns:
        print(c["name"], "| Reach:", c["reach"],
              "| Conversions:", c["conversions"], "| Revenue: Rs.", c["revenue"])

def product_menu():
    while True:
        print("\n--- PRODUCTS ---")
        print("1. View  2. Add  3. Search  4. Update  5. Back")
        choice = input("Choice: ")
        if choice == "1":
            show_products()
        elif choice == "2":
            add_product()
        elif choice == "3":
            search_product()
        elif choice == "4":
            update_product()
        elif choice == "5":
            break
        else:
            print("Invalid choice.")

def campaign_menu():
    while True:
        print("\n--- CAMPAIGNS ---")
        print("1. View  2. Create  3. Search  4. Update status")
        print("5. Delete  6. Record performance  7. Back")
        choice = input("Choice: ")
        if choice == "1":
            show_campaigns()
        elif choice == "2":
            add_campaign()
        elif choice == "3":
            search_campaign()
        elif choice == "4":
            update_campaign()
        elif choice == "5":
            delete_campaign()
        elif choice == "6":
            record_performance()
        elif choice == "7":
            break
        else:
            print("Invalid choice.")

def customer_menu():
    while True:
        print("\n--- CUSTOMERS ---")
        print("1. Register  2. View all  3. Search  4. Back")
        choice = input("Choice: ")
        if choice == "1":
            add_customer()
        elif choice == "2":
            show_customers()
        elif choice == "3":
            search_customer()
        elif choice == "4":
            break
        else:
            print("Invalid choice.")

def promotion_menu():
    while True:
        print("\n--- PROMO CODES ---")
        print("1. Create  2. View  3. Activate/Deactivate  4. Back")
        choice = input("Choice: ")
        if choice == "1":
            add_promotion()
        elif choice == "2":
            show_promotions()
        elif choice == "3":
            toggle_promotion()
        elif choice == "4":
            break
        else:
            print("Invalid choice.")

def main():
    print("=" * 40)
    print("WELCOME TO FRIZO MANAGEMENT PORTAL")
    print("Fries. Your Way.")
    print("=" * 40)

    if not login():
        return

    while True:
        show_dashboard()
        print("\n===== MAIN MENU =====")
        print("1. Products")
        print("2. Campaigns")
        print("3. Customers")
        print("4. Promo Codes")
        print("5. Reports")
        print("6. Exit")
        choice = input("Choice: ")
        if choice == "1":
            product_menu()
        elif choice == "2":
            campaign_menu()
        elif choice == "3":
            customer_menu()
        elif choice == "4":
            promotion_menu()
        elif choice == "5":
            show_report()
        elif choice == "6":
            print("Thank you for using FRIZO!")
            break
        else:
            print("Invalid choice. Select 1 to 6.")

main()
```
